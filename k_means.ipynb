```python
import numpy as np
```

```python
X = np.array([
    # Cluster 1
    [1, 1],
    [1.5, 2],
    [2, 1],
    [2, 2],
    [1, 1.5],

    # Cluster 2
    [8, 8],
    [9, 8],
    [8, 9],
    [9, 9],
    [8.5, 8.5],

    # Cluster 3
    [4, 10],
    [5, 11],
    [4.5, 10.5],
    [5, 10],
    [4, 11]
])
```

```python
from k_means import KMeans
model = KMeans(
    k=3,
    max_iters=100,
    random_state=42
)

model.fit(X)
```

```python
print(model.labels_)
```

```python
print(model.centroids)
```

```python
print(model.inertia_)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))

for cluster in range(model.k):

    points = X[model.labels_ == cluster]

    plt.scatter(
        points[:, 0],
        points[:, 1],
        label=f"Cluster {cluster}"
    )

plt.scatter(
    model.centroids[:, 0],
    model.centroids[:, 1],
    marker="X",
    s=200,
    label="Centroids"
)

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("K-Means Clustering")
plt.legend()
plt.show()
```

```python
inertias = []

for k in range(1, 8):

    model = KMeans(
        k=k,
        random_state=42
    )

    model.fit(X)

    inertias.append(model.inertia_)
```

```python
plt.plot(
    range(1, 8),
    inertias,
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")

plt.show()
```

```python
from sklearn.cluster import KMeans as SKKMeans

sk_model = SKKMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

sk_model.fit(X)
```

```python
print(sk_model.labels_)
```

```python
print(sk_model.cluster_centers_)
```

```python
print(sk_model.inertia_)
```
