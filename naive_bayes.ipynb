## Algorithm Steps

Our classifier will work like this:

Training Data
     ↓
Separate samples by class
     ↓
Calculate class prior
     ↓
For each class:
    Calculate mean of each feature
    Calculate variance of each feature
     ↓
Store statistics

For prediction:

New sample
     ↓
Calculate Gaussian probability
for every feature
     ↓
Multiply feature probabilities
     ↓
Multiply by class prior
     ↓
Repeat for every class
     ↓
Choose class with highest probability


	​


```python
import numpy as np
X = np.array([
    [1.0, 1.2],
    [1.2, 1.0],
    [0.8, 1.1],
    [1.1, 0.9],

    [5.0, 5.2],
    [5.2, 5.0],
    [4.8, 5.1],
    [5.1, 4.9]
])

y = np.array([
    0,
    0,
    0,
    0,

    1,
    1,
    1,
    1
])
```

```python
from naivebayes import GaussianNaiveBayes
model = GaussianNaiveBayes()

model.fit(X, y)
```

```python
X_test = np.array([
    [1.0, 1.0],
    [5.0, 5.0],
    [3.0, 3.0]
])

predictions = model.predict(X_test)

print(predictions)
```

```python
import matplotlib.pyplot as plt

plt.scatter(
    X[:, 0],
    X[:, 1],
    c=y
)

plt.scatter(
    X_test[:, 0],
    X_test[:, 1],
    marker="x",
    s=100
)

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Gaussian Naive Bayes")

plt.show()
```

```python
def accuracy_score(y_true, y_pred):
    return np.mean(y_true == y_pred)
```

```python
y_pred = model.predict(X)

print(
    "Accuracy:",
    accuracy_score(y, y_pred)
)
```

```python
from sklearn.naive_bayes import GaussianNB

sk_model = GaussianNB()

sk_model.fit(X, y)

sk_predictions = sk_model.predict(X_test)

print("Our model:")
print(predictions)

print("Sklearn:")
print(sk_predictions)
```
